```python
import gradio as gr
```

```python
def sarvam_chat(user_message):
    response = client.chat.completions(
        model="sarvam-m",
        messages=[
            {"role": "system", "content": "You are a knowledgeable and helpful Real Estate Investment Advisor. Your task is to provide advice, insights, and answer questions related to real estate investments. Focus on market analysis, property types, investment strategies, and potential returns. Always emphasize the importance of due diligence, understanding market risks, considering individual financial situations, and seeking professional legal or financial advice. Provide general guidance and insights, but do not offer specific investment recommendations."},
            {"role": "user", "content": user_message}
        ]
    )
    return response.choices[0].message.content
```

```python
def sarvam_chat_with_tokens(user_message_1, user_message_2):
    # Process the first message
    response_1 = client.chat.completions(
        model="sarvam-m",
        messages=[
            {"role": "system", "content": "You are a knowledgeable and helpful Real Estate Investment Advisor. Your task is to provide advice, insights, and answer questions related to real estate investments. Focus on market analysis, property types, investment strategies, and potential returns. Always emphasize the importance of due diligence, understanding market risks, considering individual financial situations, and seeking professional legal or financial advice. Provide general guidance and insights, but do not offer specific investment recommendations."},
            {"role": "user", "content": user_message_1}
        ]
    )
    content_1 = response_1.choices[0].message.content
    tokens_1 = response_1.usage.total_tokens if response_1.usage else 'N/A'

    # Process the second message
    response_2 = client.chat.completions(
        model="sarvam-m",
        messages=[
            {"role": "system", "content": "You are a knowledgeable and helpful Real Estate Investment Advisor. Your task is to provide advice, insights, and answer questions related to real estate investments. Focus on market analysis, property types, investment strategies, and potential returns. Always emphasize the importance of due diligence, understanding market risks, considering individual financial situations, and seeking professional legal or financial advice. Provide general guidance and insights, but do not offer specific investment recommendations."},
            {"role": "user", "content": user_message_2}
        ]
    )
    content_2 = response_2.choices[0].message.content
    tokens_2 = response_2.usage.total_tokens if response_2.usage else 'N/A'

    return f"Response 1:\n{content_1}\n\nTotal tokens (Prompt 1): {tokens_1}", \
           f"Response 2:\n{content_2}\n\nTotal tokens (Prompt 2): {tokens_2}"
```

```python
iface = gr.Interface(fn=sarvam_chat, inputs=gr.Textbox(lines=5, label="Your Real Estate Investment Question"), outputs=gr.Textbox(lines=15, label="Real Estate Investment Advice"), title="Real Estate Investment Advisor")
iface.launch(debug=True)
```

```python
iface_two_prompts = gr.Interface(
    fn=sarvam_chat_with_tokens,
    inputs=[
        gr.Textbox(lines=5, label="Your First Real Estate Investment Question"),
        gr.Textbox(lines=5, label="Your Second Real Estate Investment Question")
    ],
    outputs=[
        gr.Textbox(lines=15, label="Response 1 (with Tokens)"),
        gr.Textbox(lines=15, label="Response 2 (with Tokens)")
    ],
    title="Real Estate Investment Advisor (Dual Prompts with Token Count)"
)

iface_two_prompts.launch(debug=True)
```
